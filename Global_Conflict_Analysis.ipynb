## **Feature Engineering**

```python
df['decade'] = (df['year'] // 10) * 10
```

```python
df['is_deadly'] = df['killed'].apply(lambda x: 'Yes' if x > 10 else 'No')
```

## **Visualization**


```python
df
```

```python
# Incidents by region and attack type
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

df['region'].value_counts().plot(kind='barh', ax=axes[0], color='steelblue', edgecolor='white')
axes[0].set_title('Incidents by Region')
axes[0].set_xlabel('Count')

df['attack_type'].value_counts().plot(kind='bar', ax=axes[1], color='coral', edgecolor='white')
axes[1].set_title('Incidents by Attack Type')
axes[1].set_xlabel('Attack Type')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Avg casualties by attack type
plt.figure(figsize=(9, 4))
atk_cas = df.groupby('attack_type')['killed'].mean().sort_values(ascending=False)
atk_cas.plot(kind='bar', color='tomato', edgecolor='white')
plt.title('Average Killed per Attack Type')
plt.xlabel('Attack Type')
plt.ylabel('Avg Killed')
plt.xticks(rotation=40, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Success rate by attack type
success = df.groupby('attack_type')['success'].mean().sort_values(ascending=False)

plt.figure(figsize=(9, 4))
success.mul(100).plot(kind='bar', color='mediumseagreen', edgecolor='white')
plt.title('Success Rate (%) by Attack Type')
plt.xlabel('Attack Type')
plt.ylabel('Success Rate (%)')
plt.xticks(rotation=40, ha='right')
plt.axhline(y=success.mean()*100, color='red', linestyle='--', linewidth=1, label='Overall avg')
plt.legend()
plt.tight_layout()
plt.show()
```

##  Key Insights — Global Conflict Analysis

1. **Middle East and Africa** are the most conflict-affected regions globally.
2. **Bombing/Explosion** is the most common and deadliest attack type by average casualties.
3. **Most attacks succeed** — overall success rate is ~73%, meaning defence/prevention is difficult.
4. **Hijacking and Assassination** have lower success rates — more complex to execute.
5. **Civilians and Military** are the most frequent targets across all regions.
